```python
import pandas as pd

df = pd.read_csv("./data/weather/65 Years of Weather Data Bangladesh (1948 - 2013).csv")

print("DataFrame Shape:", df.shape)
print("\nDataFrame Types:")
print(df.dtypes)
print("\nMissing Values:")
print(df.isna().sum())
```

```python
'''65 years claimed to be from 1948 to 2013'''

years = df["YEAR"].unique()
print("Unique Years: ", len(years), "from", years.min(), "to", years.max())
total_months = years.size * 12
print("Total Months: ", total_months)

stations = df["Station Names"].unique()
print("Unique Stations: ", len(stations))
# print("Station Names: ", stations)
print("\nRows Per Station:\n", df["Station Names"].value_counts())
```

```python
coverage_per_stations = df["Station Names"].value_counts()
# print("\nCoverage Per Station:\n", coverage_per_stations)

full_coverage_stations = coverage_per_stations[coverage_per_stations == total_months]
print("\nStations with Full Coverage ({} months): {}\n".format(total_months, len(full_coverage_stations)))

zeros = df[df["ALT"] == 0]
# print("Rows with Zero Altitude: ", zeros)
print("Rows with Zero Altitude: ", len(zeros))

print("\nStations with Zero Altitude:\n", df.loc[df["ALT"] == 0, "Station Names"].value_counts())
```

```python
# Stations' value of station number, X_cor, Y_cor, Laitude, Logitude, Altitude, and Station Names. These are constant for each station.

unique_station_names = df["Station Names"].unique()
# print("\nUnique Station Names: ", unique_station_names)

print(f"{'Station':<25} | {'Number':<8} | {'X Coord':<12} | {'Y Coord':<12} | {'Latitude':<10} | {'Longitude':<10} | {'Altitude':<8}")
print("-" * 105) # বিভাজক রেখা

for station in df["Station Names"].unique():
    station_data = df[df["Station Names"] == station].iloc[0]
    
    print(
        f"{str(station):<25} | "
        f"{str(station_data['Station Number']):<8} | "
        f"{str(station_data['X_COR']):<12} | "
        f"{str(station_data['Y_COR']):<12} | "
        f"{str(station_data['LATITUDE']):<10} | "
        f"{str(station_data['LONGITUDE']):<10} | "
        f"{str(station_data['ALT']):<8}"
    )
```

```python
excluded_stations = ["Chittagong (City-Ambagan)", "Chuadanga", "Syedpur"]

# Rows for each excluded station
for station in excluded_stations:
    station_rows = df[df["Station Names"] == station]
    print(f"\nRows for {station}: {len(station_rows)}")
    print(f"Years: {station_rows['YEAR'].min()} to {station_rows['YEAR'].max()}")
```

```python
# I want to check Chittagong (City-Ambagan) station's data for the years 2009 to 2013 side by side. Years will be the columns and the weather parameters will be the rows. This will help me to see if there are any missing values or anomalies in the data for this station during these years.

chittagong_data = df[(df["Station Names"] == "Chittagong (City-Ambagan)") & (df["YEAR"].between(2009, 2013))]
weather_cols = [
    "Max Temp", 
    # "Min Temp", 
    # "Rainfall", 
    # "Relative Humidity",
    # "Wind Speed", 
    # "Cloud Coverage", 
    # "Bright Sunshine"
]

chittagong_data_pivot = (
    chittagong_data
    .melt(
        id_vars=["Month", "YEAR"],
        value_vars=weather_cols,
        var_name="Weather Parameter",
        value_name="Value"
    )
    .pivot_table(
        index=["Weather Parameter", "Month"],
        columns="YEAR",
        values="Value",
        aggfunc="first"
    )
    .sort_index()
)
print(chittagong_data_pivot)
```

```python
# I want to check Chuadanga station's data for the years 1989 to 2013 side by side. Years will be the columns and the weather parameters will be the rows. This will help me to see if there are any missing values or anomalies in the data for this station during these years.

chuadanga_data = df[(df["Station Names"] == "Chuadanga") & (df["YEAR"].between(1989, 2013))]
weather_cols = [
    "Max Temp", 
    # "Min Temp", 
    # "Rainfall", 
    # "Relative Humidity",
    # "Wind Speed", 
    # "Cloud Coverage", 
    # "Bright Sunshine"
]

chuadanga_data_pivot = (
    chuadanga_data
    .melt(
        id_vars=["Month", "YEAR"],
        value_vars=weather_cols,
        var_name="Weather Parameter",
        value_name="Value"
    )
    .pivot_table(
        index=["Weather Parameter", "Month"],
        columns="YEAR",
        values="Value",
        aggfunc="first"
    )
    .sort_index()
)
print(chuadanga_data_pivot)
```
